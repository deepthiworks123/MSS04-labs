```python

import math

# Solve ax² + bx + c = 0
def q_solve(a, b, c):

    # Compute the discriminant
    D = b**2 - 4*a*c

    if D >= 0:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)

        return x1, x2

    else:
        print("no real solution")
        return None, None


x1, x2 = q_solve(1, 4, 1)

print("x1 =", x1)
print("x2 =", x2)

x1, x2 = q_solve(1, 1, 1)
```

```python
import math

# Complex solutions allowed
# Solve ax² + bx + c = 0

def q_solve(a, b, c):

    # Compute the discriminant
    D = b**2 - 4*a*c

    if D >= 0:
        sq_D = math.sqrt(D)

    else:
        sq_D = math.sqrt(-D) * 1j

    # Calculate solutions
    x1 = (-b + sq_D) / (2*a)
    x2 = (-b - sq_D) / (2*a)

    return x1, x2


# Test the function
x1, x2 = q_solve(1, 4, 1)

print("x1 =", x1)
print("x2 =", x2)

x1, x2 = q_solve(1, 1, 1)

print("x1 =", x1)
print("x2 =", x2)
```
